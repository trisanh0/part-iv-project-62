```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tsfresh
from tsfresh import extract_relevant_features, select_features
from tsfresh.feature_extraction import extract_features
from tsfresh.utilities.dataframe_functions import impute
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.model_selection import RepeatedStratifiedKFold, cross_val_score
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("X_train.txt", sep=r"\s+", header=None)
df["id"] = df.index
df = df.melt(id_vars="id", var_name="time", value_name= "value")
df_test = pd.read_csv("X_test.txt", sep=r"\s+", header=None)
df_test["id"] = df_test.index
df_test = df_test.melt(id_vars="id", var_name="time", value_name="value")
y = pd.read_csv("y_train.txt", header = None).squeeze()
y_test = pd.read_csv("y_test.txt", header=None).squeeze()
# print(df)
# print(y)
print(df_test)
```

```python
X = extract_features(df, column_id="id", column_sort="time")
X = X.dropna(axis=1)
X_test = extract_features(df_test, column_id="id", column_sort="time")
X_test = X_test.dropna(axis=1)
```

```python
X_selected = select_features(X, y)
```

```python
X_test_selected = X_test[X_selected.columns]
y_test_aligned = y_test.reindex(X_test_selected.index)

model = GradientBoostingClassifier(
    n_estimators=20,
    max_depth=3
)

model.fit(X_selected, y)

y_pred = model.predict(X_test_selected)

print("Test accuracy:", accuracy_score(y_test_aligned, y_pred))
```
